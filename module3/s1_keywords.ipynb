# Extraction de Keywords

## Imports

```python
import os
import yake
```

## Extraire les mots clés d'un document avec Yake

https://github.com/LIAAD/yake
https://github.com/INESCTEC/yake

```python
# Instantier l'extracteur de mots clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50) #on aurait du utiliser un autre type dde ngram, à noter que, dans les paramètres initaux (dans la documentation) de ce tp, le ngram est définit par défaut à 3gram, mais on peut changer ça ici, entre les parenthèse on ajoute (...,n=4)
kw_extractor
```

```python
# Lister les Fichiers dont le nom contient 1963 
data_path = "../data/txt/"
year = "1963"

files = [
    f for f in os.listdir(data_path)
    if f.endswith(".txt") and year in f
]
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

```python
# Les dix premiers fichiers
files[:10]
```

```python
# Choisir un fichier
this_file = files[0]
this_file
```

```python
# Récupérer le texte du fichier
text = open(os.path.join(data_path, this_file), 'r', encoding='utf-8').read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords #par defaut, il prend les 3grammes, mais on peut les changer dans les paramètres en haut
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
for f in sorted(files)[:]:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2: #ici nous avons pris que les bi-grammes
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```
