## Utiliser [OCR.space](https://ocr.space/ocrapi) pour convertir un PDF (contenant une image scannée) en texte exploitable

```python
## clé API et URL du document
API_KEY = "K8935297046716088957" #j'ai eu cet API Key après inscription (reçu par email)
PDF_URL = "https://max.de.wilde.web.ulb.be/camille/KB_JB92_1873-02-06_01-00002.pdf" #j'ai choisi un fichier pdf du sous-ensemble Camille (au hasard) pour effectuer ce test 

## requête vers l'API OCR.space
url = "https://api.ocr.space/parse/image"

data = {
    "apikey": API_KEY,
    "url": PDF_URL,
    "language": "fre", #j'ai consulté le document avant de le mettre ici. La langue était le français
    "OCREngine": "2" #d'après la documentation et les recherches en ligne, le premier engin a été retiré et redirigé vers le moteur 2
}

resp = requests.post("https://api.ocr.space/parse/image", data=data)
result = resp.json()

text = result["ParsedResults"][0]["ParsedText"]
print(text)
```
