# Sentiment analysis 

## 1. Textblob-FR

Documentation: https://textblob.readthedocs.io/en/dev/

### Imports

```python
from textblob import Blobber
from textblob_fr import PatternTagger, PatternAnalyzer
```

### Création d'une fonction `get_sentiment`

```python
tb = Blobber(pos_tagger=PatternTagger(), analyzer=PatternAnalyzer())

def get_sentiment(input_text):
    blob = tb(input_text)
    polarity, subjectivity = blob.sentiment
    polarity_perc = f"{100*abs(polarity):.0f}"
    subjectivity_perc = f"{100*subjectivity:.0f}"
    if polarity > 0:
        polarity_str = f"{polarity_perc}% positive"
    elif polarity < 0:
        polarity_str = f"{polarity_perc}% negative"
    else:
        polarity_str = "neutral"
    if subjectivity > 0:
        subjectivity_str = f"{subjectivity_perc}% subjective"
    else:
        subjectivity_str = "perfectly objective"
    print(f"This text is {polarity_str} and {subjectivity_str}.")
```

### Analyser le sentiment d'une phrase

```python
sentences_1963 = [
    "éviter une mascarade que le grand public n'apprécie plus",
    "Messieurs les actionnaires et porteurs de parts de fondateur sont priés d'assister A l'Assemblée générale extraordinaire qui se tiendra le 11 février 1963.",
    "Les intempéries avaient perturbé le marché",
    "Un lecteur suggérait, l’autre jour, de fleurir nos avenues et nos fenêtres afin de rendre la future capitale de l'Europe souriante à nos visiteurs.",
    "Cette année, après quatre matches de championnat, elle occupe la dernière place du classement6. vous connaissez le français et le flamand.",
    "Les précisions du détenu ont permis à la police de faire un dessin- robot de la jeune femme et de son compagnon.",
    "Ce dernier aurait même été identifié.",
    "On est encore relativement éloigné du boom de Corée.",
    "Ce joueur, aux qualités affirmées, fut victime d’un accident et dut être opéré du ménisque.",
    "Les flammes furent éteintes rapidement."
]
```

Je vais analyser le sentiment de chacune des phrases, en suivant le même ordre ci-dessus:

```python
results = []

for i, sentence in enumerate(sentences_1963, start=1):

    get_sentiment(sentence)
    
```
